```python
products_df = pd.read_csv("../data/products.csv")

customers_df = pd.read_csv("../data/customers.csv")

orders_df = pd.read_csv("../data/orders.csv")

returns_df = pd.read_csv("../data/returns.csv")

print("Datasets loaded successfully.")
```

```python
# BASIC CLEANING
```

```python
# Remove Duplicates

orders_df = orders_df.drop_duplicates()

customers_df = customers_df.drop_duplicates()

products_df = products_df.drop_duplicates()

returns_df = returns_df.drop_duplicates()

print("Duplicates removed.")
```

```python
# Convert Dates

orders_df["Order_Date"] = pd.to_datetime(
    orders_df["Order_Date"],
    errors="coerce"
)

orders_df["Ship_Date"] = pd.to_datetime(
    orders_df["Ship_Date"],
    errors="coerce"
)

customers_df["Join_Date"] = pd.to_datetime(
    customers_df["Join_Date"],
    errors="coerce"
)

print("Date conversion completed.")
```

```python
# Fill Missing Discounts

orders_df["Discount"] = (
    orders_df["Discount"]
    .fillna(0)
)
```

```python
# DERIVED METRICS
```

```python
# Shipping Days
orders_df["Shipping_Days"] = (
    orders_df["Ship_Date"]
    - orders_df["Order_Date"]
).dt.days

# Profit Margin %
orders_df["Profit_Margin_Percent"] = (
    orders_df["Profit"]
    / orders_df["Sales"]
) * 100
```

```python
# MASTER ANALYTICS DATAFRAME
```

```python
# =====================================================
# MERGE DATASETS
# =====================================================

master_df = orders_df.merge(
    customers_df,
    on="Customer_ID",
    how="left"
)

master_df = master_df.merge(
    products_df,
    on="Product_ID",
    how="left"
)

master_df = master_df.merge(
    returns_df,
    on="Order_ID",
    how="left"
)

print("Master DataFrame created successfully.")
```
